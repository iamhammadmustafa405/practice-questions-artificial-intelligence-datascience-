```python
# 3: Find the second largest number in a list without sorting.

def second_largest_num(list):
    max = list[0]
    second_largest = None
    for idx in range(len(list)):
        if list[idx] > max:
            second_largest = max
            max = list[idx]

    return second_largest

l= [4,2,3,6,8,5,9]
second_largest_num(list= l)
second_largest_num(list= [2,3,4,5])
```

```python
# 5: Find common elements between two lists.
l= [2,3,7,9,5]
m = [2,7,6,10,8]

def common_elements(list1,list2):
    common = []

    for i in list1:
        for j in list2:
            if i == j:
                common.append(i)
    return common

common_elements(list1=l,list2 = m)
common_elements(list1=[1,2,3,4,5,6,7],list2= [2,3,5,6,7])
```

```python
# 6: Create a dictionary from two lists.
key = ["name","marks","age"]
value = ["Hammad",90,19]

def dict_from_lists(keys,values):

    d= {}
    for i in range(len(keys)):
        d[keys[i]] = values[i]
    return d

dict_from_lists(keys= key, values= value)
dict_from_lists(keys=["Id","Name","Rollno","grade"],values= [1,"Hammad Mustafa",3,"A"])
```

```python
# 7: Invert a dictionary's keys and values.
def invert_dict(dict):
    keys= list(dict.keys())
    values = list(dict.values())
    d_invert = {}
    for i in range(len(keys)):
        d_invert[values[i]] = keys[i]
    return d_invert


invert_dict({"Hello": 560})
```

```python
# 8: Check whether a string is a palindrome while ignoring spaces and case.

def check_palindrome(str):
    if str.lower().strip() == str[::-1].lower().strip():
        return True
    else:
        return False

check_palindrome("  LAtal    ")
```

```python
# 9: Write a program that takes the user's height in centimetres and weight in kilograms and calculates
# their BMI (BMI = weight ÷ height² in metres). Then print whether they are: Underweight (<18.5),
# Normal (18.5–24.9), Overweight (25–29.9), or Obese (≥30).


def weight_checker(height_cm,weight_kg):
    bmi_m= (weight_kg/(height_cm**2))
    print(bmi_m)
    if bmi_m < 18.5:
        return "Underweight"
    elif (bmi_m >=18.5 and bmi_m <=24.9):
        return "Normal"
    elif (bmi_m >=25 and bmi_m <=29.9):
        return "Overweight"
    elif bmi_m >30:
        return "Obese"

weight_checker(height_cm= 5.5,weight_kg=90)
```
