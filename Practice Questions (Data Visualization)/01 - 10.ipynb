```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go

np.random.seed(42) # reproducible data
```

```python
# 1: Plot sales against months as a line chart. Add a title "Monthly Sales", an x-axis label
# "Month", and a y-axis label "Units Sold".
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 135, 128, 160, 175, 190]

plt.plot(months,sales,linewidth=2,color="red",marker="o",linestyle="dotted")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Units Sold")
```

```python
# 2: Draw a bar chart of revenue per category. Give each bar a different color and label the
# axes.
categories = ["Electronics", "Clothing", "Food", "Toys", "Books"]
revenue = [45, 30, 55, 20, 25]

plt.figure(figsize=(10,10))
sns.barplot(x=categories,y=revenue,hue=categories)

```

```python
# using plotly
px.bar(x=categories,y=revenue,color=categories)
```
