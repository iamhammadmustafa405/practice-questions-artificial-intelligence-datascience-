```python
# Q11. Find the total revenue for each category (Electronics, Clothing, Books). Total revenue =
# quantity × price. Which category earns the most?

# My logic
def revenue(group):
    group["revenue"] = group["quantity"] * group["price"]
    return  group["revenue"].sum()


df.groupby("category").apply(revenue).sort_values(ascending=False) # part 01
```

```python
catg = df.groupby("category").apply(revenue).sort_values(ascending=False).head(1).index[0] # part 2
print(f"The Category which earns the most is: {catg}")
```

```python
# Actual method:
df["revenue"] = df["quantity"] * df["price"]
df.groupby("category")["revenue"].sum().sort_values(ascending = False) # 1st part
```

```python
catg1 = df.groupby("category")["revenue"].sum().sort_values(ascending = False).head(1).index[0] # part 2
print(f"The Category which earns the most is: {catg1}") 
```

```python
# Q12. Show only the orders that have a discount greater than 0 AND belong to the 'Electronics'
# category.
# 1st method
df[(df["discount_pct"] > 0) & (df["category"] == "Electronics")]
```

```python
# 2nd method
electronics = df.groupby("category").get_group("Electronics")
electronics[electronics["discount_pct"] >0]
```

```python
# Q13. Fill the missing values in the 'rating' column with the average rating. Do this on a copy of the
# DataFrame so the original is unchanged.
df_copy = df.copy()


df_copy["rating"] = df_copy["rating"].fillna(np.round(df_copy["rating"].mean(),2))
df_copy
```

```python
# Q14. Add a new column called 'final_price' which is the price after applying the discount. Formula:
# final_price = price × (1 - discount_pct / 100).
df["final_price"]= df["price"] * (1 - (df["discount_pct"]/100))
df.head(3)
```

```python
# Q15. Find the top 3 customers who spent the most money in total. (Use quantity × price as
# spending.)
df.sort_values("revenue",ascending= False)[["customer_name","revenue"]].rename(columns={"revenue": "spending"}).head(3)
```

```python
# Q16. Extract the month number from 'order_date' into a new column called 'month'. Then count
# how many orders were placed each month.
df["month"] = df["order_date"].dt.month
df.groupby("month")["customer_name"].count()
```

```python
# Q17. Show all orders where the status is either 'Cancelled' or 'Pending'.
df[(df["status"] == "Cancelled") | (df["status"] =="Pending")] # 1st method
```

```python
df[df["status"].isin(["Cancelled","Pending"])] # using isin() --> 2nd method
```

```python
# Q18. Find the average rating for each product category, but ignore the missing ratings (NaN) when
# calculating.  
np.round(df.groupby("category")["rating"].mean(skipna=True),2) # skipna = True (Default)
```

```python
# Q19. Create a new column 'is_expensive' that is True if the price is above 5000, and False
# otherwise. Then count how many expensive vs non-expensive products there are.
df["is_expensive"] = df["price"] > 5000
df.groupby("is_expensive")["customer_name"].count().reset_index()  # 1st method
```
