```python
# Q29. Rank customers by their total spending (quantity × price). Add a column 'spending_rank'
# where rank 1 = highest spender. Customers with equal spending should share the same rank.
df["total_spending"] = df["quantity"] * df["price"]
df["spending_rank"] = df["total_spending"].rank(method= "min", ascending = False)
df.sort_values("spending_rank")
```

```python
# Q30. For each order, calculate the cumulative revenue over time (sorted by order_date). Add this
# as a column 'cumulative_revenue'
df["cumulative_revenue"] = np.cumsum(df["total_spending"])
df
```

```python
# Q31. Find which city has the highest cancellation rate (cancelled orders ÷ total orders × 100)
total_orders = df.shape[0]
cancelled_df = df[df["status"] =="Cancelled"]
cancelled_order = cancelled_df.groupby("city")["status"].count()

cancelled_per = (cancelled_order/total_orders) * 100
cancelled_per # each city has 1 cancelled order and rate is 5% so all these are has highest cancellation rate
```

```python
# Q32. Use .apply() to create a custom function that gives a 'performance label' to each order: if
# rating >= 4.5 → 'Excellent', 3.5–4.4 → 'Good', below 3.5 → 'Needs Improvement', NaN → 'Not
# Rated'.

def performance(rating):
    if rating  >=4.5:
        return "Excellent"
    elif (rating >=3.5) & (rating <=4.4):
        return "Needs Improvement"
    elif (rating == np.nan) | (rating == None):
        return "Not Rated"


df["performance_label"] = df["rating"].apply(performance)
df
```

```python
# Q33. Reshape the data: create a wide-format table where each row is a customer, each column is
# a product category, and the values are the total amount spent by that customer in that category.
# Fill missing combinations with 0.

# 1st method using pivot_table()
df.pivot_table(index="customer_name",columns="category",values="total_spending").fillna(0)
```

```python
# 2nd method ---> by using groupby,unstack() and set_index
df.set_index(["customer_name","category"]).groupby(["customer_name","category"])["total_spending"].sum().unstack().fillna(0)
```

```python
# Q34. For each customer, find the time gap (in days) between their first and last order. Show the
# result sorted from the customer with the longest gap to the shortest.


# in this question timedelta concept is used that i didn't cover yet.

first_order_date = df.groupby("customer_name")["order_date"].min()
last_order_date = df.groupby("customer_name")["order_date"].max()
(last_order_date - first_order_date).reset_index().rename(columns= {"order_date":"time_gap"}).sort_values("time_gap",ascending = False)


```

```python
# in one line
((df.groupby("customer_name")["order_date"].max()) - (df.groupby("customer_name")["order_date"].min())).reset_index().rename(columns={"order_date": "time_gap"}).sort_values("time_gap",ascending = False)
```

```python
# Q35. Build a full summary report in one chained pipeline (no intermediate variables). The output
# should show: city, total orders, total revenue, avg rating (rounded to 2), and cancellation rate (%)
# — all in one DataFrame, sorted by total revenue descending

# Chained Pipeline

df.groupby("city")[["quantity","total_spending","rating",]].agg(func= {"quantity":"count","total_spending":"sum","rating":"mean",}).rename(columns = {"total_spending":"total_revenue","quantity":"total_orders","rating":"avg_rating"}).reset_index().merge(((df[df["status"] =="Cancelled"].groupby("city")["quantity"].count()/df.shape[0]) * 100).reset_index().rename(columns={"quantity":"cancellation_rate_%"})).sort_values("total_revenue",ascending = False)
```
