```python
# Q23. Create a pivot table that shows the total revenue (quantity × price) for each city (rows) and
# each category (columns).
df["revenue"] = df["quantity"] * df["price"]
df.pivot_table(index=["city"],columns=["category"],values= ["revenue"])
```

```python
# 2nd method
df.groupby("customer_name")["customer_name"].count().sort_values(ascending = False)
```

```python
# Q25. Calculate the month-over-month total revenue. Which month had the highest revenue?
df["month"] = df["order_date"].dt.month_name()
month_rev= df.groupby("month")["revenue"].sum().sort_values(ascending = False).head(1).index[0]
print(f"The month which has the highest revenue is : {month_rev}")
```

```python
# Q26. For each customer, find their average discount percentage. Then show only customers who
# received an average discount above 10%.
avg_dct = np.round(df.groupby("customer_name")["discount_pct"].mean(),2)
avg_dct[avg_dct >10] # answer is empty series not a single customer get avg, discount >10%
pd.cut()
```

```python
# Q27. Mark orders as 'High Value' if revenue (quantity × price) is above 10,000, 'Medium Value' if
# between 3,000 and 10,000, and 'Low Value' if below 3,000. Add this as a new column called
# 'order_value_tier'.

def order_tier(revenue):
    if revenue > 10000:
        return "High Value"
    elif (revenue >= 3000) & (revenue <10000):
        return "Medium Value"
    elif revenue < 3000:
        return "Low Value"
    

df["order_value_tier"] = df["revenue"].apply(order_tier)
df
```

```python
# Q28. Find the percentage share of each category's revenue out of the total revenue. Round to 2
# decimal places.
tot_reven = df["revenue"].sum()
group_reven = df.groupby("category")["revenue"].sum() 
np.round((group_reven/tot_reven) * 100,2).reset_index().rename(columns={"revenue":"revenue_%"})
```
